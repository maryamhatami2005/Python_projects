```python
import smtplib
from email.mime.text import MIMEText
import os

password = os.getenv("PASSWORD")
```

```python
def send_email(sender, recipient, sbj, body):
    msg = MIMEText(body)
    msg['Subject'] = sbj
    msg['From'] = sender
    msg['To'] = recipient

    smtp_server = "smtp.gmail.com"
    smtp_port = 587

    server = smtplib.SMTP(smtp_server, smtp_port)
    server.starttls()
    server.login(sender, password)
    server.send_message(msg)
    print("Email sent successfully")
    server.quit()
```

```python
sender = "maryhatami2005@gmail.com"
recipient = "maryamht296@gmail.com"
sbj = "test"
body = "This is a test email sent from python"

send_email(sender, recipient, sbj, body)
```

```python
from email.mime.multipart import MIMEMultipart
from email.mime.base import MIMEBase
from email import encoders
```

```python
def send_email_attachment(sender, recipient, sbj, body, file_name):
    msg = MIMEMultipart()
    msg['Subject'] = sbj
    msg['From'] = sender
    msg['To'] = recipient

    msg.attach(MIMEText(body, "plain"))

    with open(file_name, "rb") as attachment:
        part = MIMEBase("application", "pdf")
        part.set_payload(attachment.read())

    encoders.encode_base64(part)

    part.add_header("Content-Disposition", f"attachment; filename={file_name}")
    msg.attach(part)

    smtp_server = "smtp.gmail.com"
    smtp_port = 587

    server = smtplib.SMTP(smtp_server, smtp_port)
    server.starttls()
    server.login(sender, password)
    server.send_message(msg)
    print("Email sent successfully")
    server.quit()
```

```python
sender = "maryhatami2005@gmail.com"
recipient = "maryamht296@gmail.com"
sbj = "test"
body = "This is a test email sent from python"
file_name = "document.pdf"
send_email_attachment(sender, recipient, sbj, body, file_name)
```

HTML Email 

```python
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
import smtplib

def send_html_email(sender, recipient, sbj,  html, text):
    msg = MIMEMultipart("alternative")
    msg['Subject'] = sbj
    msg['From'] = sender
    msg['To'] = recipient

    part1 = MIMEText(text, "plain")
    part2 = MIMEText(html, "html")

    msg.attach(part1)
    msg.attach(part2)

    with smtplib.SMTP("smtp.gmail.com", 587) as server:
        server.starttls()
        server.login(sender, password)
        server.send_message(msg)
        print("Email sent successfully")
        server.quit()


html_content = """
<html>
  <body>
    <h1>Hello!</h1>
    <p>This is an <b>HTML</b> email sent from <i>Python</i>.</p>
    <p>Here's a list:
      <ul>
        <li>Item 1</li>
        <li>Item 2</li>
        <li>Item 3</li>
      </ul>
    </p>
  </body>
</html>
"""

text_content = """
Hello!
This is a plain text version of the HTML email sent from Python.
Here's a list:
- Item 1
- Item 2
- Item 3
"""

```

```python
sender = "maryhatami2005@gmail.com"
recipient = "maryamht296@gmail.com"
sbj = "test"

send_html_email(sender, recipient, sbj, html_content, text_content)
```

Sending Email to Multiple Recipients

```python
def send_email_to_multiple_recipients(sender, recipients, sbj, body):
    msg = MIMEMultipart()
    msg['Subject'] = sbj
    msg['From'] = sender
    msg['To'] = ",".join(recipients)

    msg.attach(MIMEText(body, "plain"))

    smtp_server = "smtp.gmail.com"
    smtp_port = 587

    server = smtplib.SMTP(smtp_server, smtp_port)
    server.starttls()
    server.login(sender, password)
    server.send_message(msg)
    print("Email sent successfully")
    server.quit()
```
