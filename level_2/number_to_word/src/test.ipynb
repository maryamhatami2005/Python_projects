```python
def factorial(n):
    if n == 0:
        return 1
    else:
        print(f"calling factorial{n-1} from factorial {n}")
        return n * factorial(n-1)

```

```python
factorial(4)
```

```python
def fibonacci(n):
    if n== 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

```python
fibonacci(2)
```

1. Write a recursive function to compute the power of a number, `power(base, exponent)`.


```python
def power(base, exponent):
    if exponent == 1 :
        return base
    else:
        return base * power(base, exponent-1)
```

```python
power(2,1)
```

```python
power(2,2)
```

```python
power(2,3)
```

2. Write a recursive function to count the number of digits in a non-negative integer.

```python
def count(n):
    if n == 0:
        return 0; 
    else:
        return count(n//10) + 1
```

```python
count(232)
```

```python
UNDER_20 = [
    "Zero",
    "One",
    "Two",
    "Three",
    "Four",
    "Five",
    "Six",
    "Seven",
    "Eight",
    "Nine",
    "Ten",
    "Eleven",
    "Twelve",
    "Thirteen",
    "Fourteen",
    "Fifteen",
    "Sixteen",
    "Seventeen",
    "Eighteen",
    "Nineteen",
]

TENS = [
    "",
    "",
    "Twenty",
    "Thirty",
    "Forty",
    "Fifty",
    "Sixty",
    "Seventy",
    "Eighty",
    "Ninety",
]

ABOVE_100 = {
    100: "Hundred",
    1000: "Thousand",
    1000000: "Million",
    1000000000: "Billion",
}
```

```python
def num_to_word(num):
    if num < 20:
        return UNDER_20 [num]
    
    elif num  < 100 : 
        remain = num % 10
        if remain == 0 :
            return TENS[num//10]
        return TENS[num//10] + " " + UNDER_20[remain]

    pivot = max([key for key in ABOVE_100 if key <= num ])
    p1 = num // pivot
    p2 = ABOVE_100[pivot]

    if num % pivot == 0:
        return f"{num_to_word(p1)} {p2}"

    return f"{num_to_word(p1)} {p2} {num_to_word(num % pivot)}"

   
```

```python
num_to_word(1453)
```

```python
num_to_word(345424)
```
