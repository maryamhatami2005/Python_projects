````python
import requests
````

````python
headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.36'
        }
requests.get("https://www.google.com", headers=headers)
````

````python
def check_site_connectivity(url):
    try:
        headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.36'}
        response = requests.get("https://www.google.com", headers=headers, timeout= 5)
        
        if response.status_code == 200:
            return True
        return False

    except Exception as e:
        return False
````

````python
check_site_connectivity("https://www.youtube.com")
````

````python
websites_to_check=[
    "https://www.google.com",
    "https://www.youtube.com",
    "https://castbox.fm",
    "https://github.com",
    "https://www.researchgate.net"
]
````

````python
for url in websites_to_check:
    print(check_site_connectivity(url))
````

````python
import threading

threads = []

for url in websites_to_check:
    thread = threading.Thread(target=check_site_connectivity, args=(url,))
    threads.append(thread)
    thread.start()
````

````python
threads
````

````python
for thread in threads:
    thread.join()
````

Instead of manually doing:
```
import threading

thread1 = threading.Thread(...)
thread2 = threading.Thread(...)
thread3 = threading.Thread(...)
```
you can often use:
```
from concurrent.futures import ThreadPoolExecutor

with ThreadPoolExecutor(max_workers=3) as executor:
    executor.map(function, items)
```

````python
from concurrent.futures import ThreadPoolExecutor

with ThreadPoolExecutor(max_workers=10) as executor:
    results = executor.map(check_site_connectivity, websites_to_check)
````

````python
for result in results:
    print(result)
````
