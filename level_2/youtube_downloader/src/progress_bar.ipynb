```python
from tqdm import tqdm
```

```python
names = ["Susan", "Nora", "Tom", "Delia", "Luis"]
```

```python
import time

for name in tqdm(names):
    print(name)
    time.sleep(1)
```

```python
for i in tqdm(range(30)):
    time.sleep(0.1)
```

```python
from tqdm import trange

for i in trange(20):
    time.sleep(0.1)
```

```python
for name in tqdm(names, desc= "Processing names:..."):
    time.sleep(0.1)
```

```python
pbar= tqdm(names, desc= "Processing names:...")
for name in pbar:
    time.sleep(0.1)
```

```python
pbar= tqdm(names)

for name in pbar:
    #pbar.set_description(name)
    pbar.set_description(f"Processing {name}")
    time.sleep(1)
```
